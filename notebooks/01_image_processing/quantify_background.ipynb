```python
overlaps = []

for i in range(len(list(coord_sample_offset.values()))):
    for j in range(i + 1, len(list(coord_sample_offset.values()))):

        r0_a, r1_a, c0_a, c1_a = list(coord_sample_offset.values())[i]
        r0_b, r1_b, c0_b, c1_b = list(coord_sample_offset.values())[j]

        row_overlap = max(r0_a, r0_b) < min(r1_a, r1_b)
        col_overlap = max(c0_a, c0_b) < min(c1_a, c1_b)

        if row_overlap and col_overlap:
            overlaps.append((i, j))

print("Overlapping pairs:")
print(overlaps)
```

```python
signal_background_files = list(Path(path_to_save).rglob("*.pickle"))
```

```python
background_marker = {}
raw_marker = {}

for file in signal_background_files:

    filename = file.name

    if filename.endswith("_background_quantified.pickle"):
        marker_key = filename.replace("_background_quantified.pickle", "")

        with open(file, "rb") as f:
            background_marker[marker_key] = pickle.load(f)

    elif filename.endswith("_raw_quantified.pickle"):
        marker_key = filename.replace("_raw_quantified.pickle", "")

        with open(file, "rb") as f:
            raw_marker[marker_key] = pickle.load(f)
```

```python
background_markers = set(background_marker.keys())
raw_markers = set(raw_marker.keys())

markers = sorted(background_markers & raw_markers)
```

```python
df = pd.DataFrame(
    [(marker, group, value)
     for marker in raw_marker
     for group, values in [("Raw", raw_marker[marker]), ("Background", background_marker[marker])]
     for value in values],
    columns=["Marker", "Image", "Intensity Sum"]
)

# Plot mean ± SEM
fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Marker",
    y="Intensity Sum",
    hue="Image",
    errorbar="se",
    capsize=0.1,
    palette={"Raw": "steelblue", "Background": "coral"},
    ax=ax
)

# Comparisons
pairs = [
    ((marker, "Raw"), (marker, "Background"))
    for marker in raw_marker
]

annotator = Annotator(
    ax,
    pairs,
    data=df,
    x="Marker",
    y="Intensity Sum",
    hue="Image"
)

annotator.configure(
    test="t-test_paired",
    text_format="star",
    loc="outside"
)

annotator.apply_and_annotate()

sns.despine()
plt.tight_layout()
plt.show()
```
